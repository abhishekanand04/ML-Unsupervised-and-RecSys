```python
# ============================================================
# BRUTE FORCE MARKET BASKET ANALYSIS (TEACHING DEMO)
# ============================================================

from itertools import combinations

print("\n==============================")
print("STEP 0: DEFINE TRANSACTIONS")
print("==============================")

transactions = [
    {1, 2, 3},   # T1
    {1, 2},      # T2
    {1, 3},      # T3
    {1, 2, 3},   # T4
    {2, 3},      # T5
    {1, 3},      # T6
    {1, 2},      # T7
    {1},         # T8
    {2},         # T9
    {4}          # T10
]

for i, t in enumerate(transactions, start=1):
    print(f"T{i}: {t}")

D = {1, 2, 3, 4}
min_support = 3

print("\nProduct Universe D =", D)
print("Minimum Support Threshold =", min_support)


# ============================================================
print("\n==============================")
print("STEP 1: GENERATE ALL ITEMSETS (BRUTE FORCE)")
print("==============================")

all_itemsets = []
for r in range(1, len(D) + 1):
    all_itemsets.extend(combinations(D, r))

print(f"Total number of itemsets generated: {len(all_itemsets)}\n")
for itemset in all_itemsets:
    print("Itemset:", set(itemset))


# ============================================================
print("\n==============================")
print("STEP 2: COUNT SUPPORT FOR EACH ITEMSET")
print("==============================")

def count_support(itemset, transactions):
    count = 0
    itemset = set(itemset)
    for t in transactions:
        if itemset.issubset(t):
            count += 1
    return count

support_counts = {}

for itemset in all_itemsets:
    support_counts[itemset] = count_support(itemset, transactions)
    print(f"Itemset {set(itemset)} occurs {support_counts[itemset]} times")


# ============================================================
print("\n==============================")
print("STEP 3: FILTER FREQUENT ITEMSETS (SUPPORT >= MIN_SUPPORT)")
print("==============================")

frequent_itemsets = {
    itemset: count
    for itemset, count in support_counts.items()
    if count >= min_support
}

for itemset, count in frequent_itemsets.items():
    print(f"FREQUENT: {set(itemset)} -> {count}")

print("\nNotice that item {4} is missing because it occurs only once.")


# ============================================================
print("\n==============================")
print("STEP 4: OBSERVE WHAT HAPPENS TO ITEMSETS CONTAINING 4")
print("==============================")

for itemset, count in support_counts.items():
    if 4 in itemset:
        print(f"Itemset {set(itemset)} contains 4 and occurs {count} times")

print("\nKey Observation:")
print("If item {4} occurs only once,")
print("then ANY itemset containing 4 can occur AT MOST once.")
print("Hence, no superset of {4} can ever be frequent.")


# ============================================================
print("\n==============================")
print("STEP 5: APRIORI-STYLE PRUNING (LOGICAL OPTIMIZATION)")
print("==============================")

print("\nKeeping ONLY frequent single items:")

frequent_single_items = {
    itemset[0]
    for itemset, count in support_counts.items()
    if len(itemset) == 1 and count >= min_support
}

print("Frequent single items:", frequent_single_items)


print("\nGenerating candidate pairs ONLY from frequent single items:")
candidate_pairs = list(combinations(frequent_single_items, 2))

for pair in candidate_pairs:
    print("Candidate Pair:", set(pair))


print("\nCounting support for candidate pairs:")

frequent_pairs = {}
for pair in candidate_pairs:
    count = count_support(pair, transactions)
    print(f"Pair {set(pair)} occurs {count} times")
    if count >= min_support:
        frequent_pairs[pair] = count


print("\nFrequent Pairs:")
for pair, count in frequent_pairs.items():
    print(f"{set(pair)} -> {count}")


# ============================================================
print("\n==============================")
print("STEP 6: TRY GENERATING SIZE-3 ITEMSETS")
print("==============================")

candidate_triplets = list(combinations(frequent_single_items, 3))

if not candidate_triplets:
    print("No candidate triplets possible.")
else:
    for triplet in candidate_triplets:
        count = count_support(triplet, transactions)
        print(f"Triplet {set(triplet)} occurs {count} times")
        if count < min_support:
            print("-> Not frequent. Algorithm stops here.")


# ============================================================
print("\n==============================")
print("FINAL CONCLUSION")
print("==============================")

print("Brute force checks ALL combinations.")
print("Apriori logic eliminates combinations that are guaranteed to fail.")
print("A superset can NEVER be more frequent than its least frequent subset.")
print("This simple logic saves enormous computation in real-world datasets.")
```

Big picture (say this before code)

“Apriori does NOT try all combinations.
It grows combinations level by level,
and kills branches early using logic.”

Key idea you already proved:

If an itemset is infrequent, all its supersets are infrequent.

That one sentence is Apriori.

How Apriori thinks (human version)

Count single items

Throw away weak ones

Combine only strong ones to make pairs

Count pairs

Throw away weak pairs

Combine strong pairs to make triplets

Repeat until nothing survives

That’s it.
